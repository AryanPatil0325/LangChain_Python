### Libraries

```python
from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate
from langchain.messages import SystemMessage,HumanMessage,AIMessage
from dotenv import load_dotenv
```

### Load the .env file

```python
load_dotenv()
```

### Initialize Model

```python
llm = init_chat_model(model="groq:qwen/qwen3.8-27b")
```

### Without using Chaining

```python
prompt = ChatPromptTemplate.from_messages([
    {"role":"system","content":"You are a {framework} expert"},
    {"role":"user","content":"{query}"}
])

final_prompt = prompt.invoke({"framework":"LangChain","query":"Explain LCEL in simple words"})

response = llm.invoke(final_prompt)
print(response.content)
```

### Using Chaining (|)

```python
prompt = ChatPromptTemplate.from_messages([
    {"role":"system","content":"You are a {framework} expert"},
    {"role":"user","content":"{query}"}
])

chain = prompt | llm

response = chain.invoke({"framework":"LangChain","query":"Explain Chaining in simple words"})

print(response.content)
```

### Self created function as node

```python
def response_content(llm_response:AIMessage):
    return llm_response.content

my_chain = prompt | llm | response_content

response = my_chain.invoke({"framework":"LangChain","query":"Explain Langchain in simple words in one sentence."})

print(response) # here response contains a string and we cannot do response.content as its not an AIMessage object
```

### Using in-build function StrOutputParser()

```python
from langchain_core.output_parsers import StrOutputParser

new_chain = prompt | llm | StrOutputParser()

response = new_chain.invoke({"framework":"LangGraph","query":"Explain LangGraph in simple words in one sentence"})

print(response)
```
