### Invoke

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()

llm = init_chat_model(
    model="groq:qwen/qwen3.8-27b",
    max_tokens = 1024
)

response = llm.invoke("Introduce yourself in short")
print(response.content)
```

### Batch

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()

llm = init_chat_model(
    model="groq:qwen/qwen3.8-27b",
    max_tokens = 1024
)
prompt = ChatPromptTemplate.from_messages([
    {"role":"user","content":"Tell me one short fun fact about {topic}"}
])

topics = [{"topic":"space"},{"topic":"ocean"},{"topic":"sand"}]

chain = prompt | llm

responses = chain.batch(topics)
for response in responses:
    print(response.content)
    
```

### Batch with controlled execution
How many queries should be run in one cycle parallelly

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv

load_dotenv()

llm = init_chat_model(
    model="groq:qwen/qwen3.8-27b",
    max_tokens = 1024
)
prompt = ChatPromptTemplate.from_template("Tell me a short joke on {topic}")

topics = [
    {"topic":"population"},
    {"topic":"humans"},
    {"topic":"Introvet guy"},
    {"topic":"IT engineer"}
]

chain = prompt | llm | StrOutputParser()

response = chain.batch(topics,config={"max_concurrency":2})

for i,resp in enumerate(response,start=1):
    print(f"\nJoke {i}:\n {resp}")
```

### Stream

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()

llm = init_chat_model(
    model="groq:qwen/qwen3.8-27b",
    max_tokens = 1024
)

prompt = ChatPromptTemplate.from_template("Explain {topic} in one sentence")

chain = prompt | llm


for response in chain.stream({"topic":"Python"}):
    print(response.content,end="")
```
