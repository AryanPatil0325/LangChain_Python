# Temperature
Controls how much variation a model can introduce while generating a response

### Low Temperature - more predictable / consistent response
### High Temperature - varied / creative response 

### Temperature = [0-1]

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()

llm_low_temp = init_chat_model(
    model="groq:qwen/qwen3.8-27b",
    temperature = 0
)
llm_high_temp = init_chat_model(
    model="groq:qwen/qwen3.8-27b",
    temperature = 1
)

for i in range(1,4):    
    response = llm_low_temp.invoke("Give me 5 creative names for a Python automation testing tool. Just names no explaination")
    print(f"Response with low temperature - {i}")
    print(response.content)
    print()

for i in range(1,4):    
    response = llm_high_temp.invoke("Give me 5 creative names for a Python automation testing tool. Just names no explaination")
    print(f"Response with high temperature - {i}")
    print(response.content)
    print()
```

# Max tokens 
maximum amount of output tokens allowed to generate a response for that request

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()

llm_1 = init_chat_model(
    model="groq:qwen/qwen3.8-27b",
    temperature = 0,
    max_tokens = 1024
)
llm_2 = init_chat_model(
    model="groq:qwen/qwen3.8-27b",
    temperature = 0,
    max_tokens = 10
)

print("High tokens to complete the response")
response = llm_1.invoke("Explain Langchain in one sentence")
print(response.content)

print()

print("Lower tokens")
res = llm_2.invoke("Explain LangChain in one sentence")
print(res.content)
```

# Timeout (in secs)
max waiting time required for a model to generate the response

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv

load_dotenv()

llm = init_chat_model(
    model="groq:qwen/qwen3.8-27b",
    timeout = 10
)
response = llm.invoke("List 5 names of people. Just names.")
print(response.content)
```
